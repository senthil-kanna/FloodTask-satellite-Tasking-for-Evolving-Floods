```python
%cd /content/floodtask

import os
import sys
import time
import shutil
import subprocess
from pathlib import Path

project = Path("/content/floodtask")
results = project / "results"
backup = Path("/content/drive/MyDrive/floodtask_results_backup")
console_log = results / "logs" / "colab_machine_b_console.log"

(results / "logs").mkdir(parents=True, exist_ok=True)
backup.mkdir(parents=True, exist_ok=True)

cmd = [
    sys.executable, "-u", "scripts/train_all.py",
    "--parallel", "2", "--machine", "b"
]

print("Starting Machine B training...")
print("Command:", " ".join(cmd))

with open(console_log, "a", buffering=1) as log:
    proc = subprocess.Popen(
        cmd,
        cwd=project,
        stdout=subprocess.PIPE,
        stderr=subprocess.STDOUT,
        text=True,
        bufsize=1
    )

    last_backup = time.time()

    while proc.poll() is None:
        line = proc.stdout.readline()
        if line:
            print(line, end="")
            log.write(line)

        if time.time() - last_backup >= 300:
            try:
                shutil.copytree(results, backup, dirs_exist_ok=True)
                print("\nDrive backup updated.")
            except Exception as e:
                print("Backup warning:", repr(e))
            last_backup = time.time()

    # Print any remaining output
    for line in proc.stdout:
        print(line, end="")
        log.write(line)

    # Final backup
    shutil.copytree(results, backup, dirs_exist_ok=True)

print("\nTraining exit code:", proc.returncode)
print("Final backup:", backup)
```

```python
import os
import zipfile

project = "/content/floodtask"
zip_path = f"{project}/results_machine_b.zip"

with zipfile.ZipFile(zip_path, "w", zipfile.ZIP_DEFLATED) as z:
    for folder in ["results/rl", "results/logs"]:
        root_path = os.path.join(project, folder)
        if os.path.isdir(root_path):
            for root, _, files in os.walk(root_path):
                for name in files:
                    path = os.path.join(root, name)
                    z.write(path, os.path.relpath(path, project))

print("Created:", zip_path)
```
